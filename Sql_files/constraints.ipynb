```python
import sqlite3 

conn = sqlite3.connect("Student_Management.db")
cursor = conn.cursor()

print("Connection successfully established and database created")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Students(
    student_id INT UNIQUE NOT NULL,
    name TEXT NOT NULL,
    age INT CHECK(age > 18),
    email TEXT UNIQUE,
    course TEXT DEFAULT 'IT',
    fees_paid INT CHECK(fees_paid > 0))
""")

conn.commit()
print("Table created successfully")

student_valid_data = [
    (1, 'Anu', 25, 'anu@gmail.com', 'Python', 25000),
    (2, 'Ravi', 22, 'ravi@gmail.com', 'Data Analytics', 30000),
    (3, 'Priya', 28, 'priya@gmail.com', 'SQL', 20000),
    (4, 'Karthik', 24, 'karthik@gmail.com', 'Power BI', 35000),
    (5, 'Meena', 30, 'meena@gmail.com', 'Excel', 15000)
]

cursor.execute("SELECT COUNT(*) FROM Students")
count = cursor.fetchone()[0]

if count == 0:
    cursor.executemany("""
        INSERT INTO Students (student_id, name, age, email, course, fees_paid)
        VALUES (?, ?, ?, ?, ?, ?)""", student_valid_data)
    conn.commit()
    print("Data entered successfully")
else:
    print("Data already exists")


cursor.execute("SELECT * FROM Students")
result = cursor.fetchall()
print(result)
```
