```python
import sqlite3

conn = sqlite3.connect("Employee.db")
cursor = conn.cursor()

print("Connection and Database created successfully")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Departments(
    Dep_ID INT PRIMARY KEY,
    Dep_name VARCHAR(50) NOT NULL)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees(
    Emp_id INT PRIMARY KEY,
    Name VARCHAR(50),
    Salary INT,
    Dep_ID INT,

    FOREIGN KEY (Dep_ID)
        REFERENCES Departments(Dep_ID))
""")
conn.commit()
print("Tables created successfully")

dept_table = [
    (10, 'IT'),
    (20, 'HR'),
    (30, 'Finance')]

cursor.executemany("""
INSERT INTO Departments 
VALUES (?,?)""", dept_table)

emp_table = [
    (1, 'Anu', 50000, 10),
    (2, 'Ravi', 60000, 20),
    (3, 'John', 45000, 10),
    (4, 'Priya', 70000, 30),
    (5, 'David', 55000, 20)]

cursor.executemany("""
INSERT INTO Employees
VALUES (?,?,?,?)""",emp_table)

conn.commit()

print("Table values updated successfully")
```

```python
cursor.execute("""
SELECT Name
FROM Employees AS e
WHERE EXISTS (
    SELECT 1
    FROM Departments AS d
    WHERE e.Dep_ID = d.Dep_ID)
""")

result = cursor.fetchall()

print(result)

#cursor.execute("""
#SELECT Name
#FROM Employees 
#WHERE Salary > ANY(
#    SELECT Salary
#    FROM Employeesx
#    WHERE dep_ID = 10)
#""")

cursor.execute("""
SELECT Name, Salary
FROM Employees
WHERE Salary > (
    SELECT MIN(Salary)
    FROM Employees
    GROUP BY dep_ID
    HAVING dep_ID = 10)
""")

result = cursor.fetchall()
print(result)

cursor.execute("""
SELECT Name, Salary
FROM Employees
WHERE Salary > (
    SELECT MAX(Salary)
    FROM Employees
    GROUP BY dep_ID
    HAVING dep_ID = 10)
""")

result = cursor.fetchall()
print(result)

conn.close()
print("Connection is closed")
```
