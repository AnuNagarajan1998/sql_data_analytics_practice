```python
import sqlite3

conn = sqlite3.connect("student_score.db")
cursor = conn.cursor()

print("Connection established and database created successfully")

cursor.execute("""
CREATE TABLE IF NOT EXISTS student(
    name TEXT,
    score INT)
""")

conn.commit()
print("Table created successfully")

scores = [
    ('Alice', 78),
    ('Bob', 92),
    ('Charlie', 65),
    ('David', 88),
    ('Emma', 74),
    ('Frank', 95),
    ('Grace', 81),
    ('Henry', 69),
    ('Isha', 87),
    ('Jack', 73)
]
cursor.execute("SELECT COUNT(*) FROM student")
count = cursor.fetchone()[0]

if count == 0:
    cursor.executemany("""
        INSERT INTO student (name, score)
        VALUES (?,?)""", scores)
    conn.commit()
    print("Data entered successfully")
else:
    print("Data already exists")


cursor.execute("""
SELECT *
FROM student
ORDER BY score ASC
LIMIT 3
""")

result = cursor.fetchall()
for r in result:
    print(r)

cursor.execute("""
SELECT *
FROM student
ORDER BY score DESC
LIMIT 3
""")

print()
result = cursor.fetchall()
for r in result:
    print(r)

#FROM row 4 to 6
cursor.execute("""
SELECT *
FROM student
ORDER BY score DESC
LIMIT 3 OFFSET 3
""")

print()
result = cursor.fetchall()
for r in result:
    print(r)

#FROM row 3
cursor.execute("""
SELECT *
FROM student
ORDER BY score DESC
LIMIT 1 OFFSET 2
""")

print()
result = cursor.fetchall()
for r in result:
    print(r)

cursor.execute("""
SELECT *,
        ROW_NUMBER() OVER (ORDER BY score DESC) AS rn
FROM student
""")

print()
result = cursor.fetchall()
for r in result:
    print(r)
conn.close()
```
